## Evaluation with Raku

```raku
my $cmd= 'numerically integrate 1/sqrt(x) over the range [0,1] without singularity handling with up to 20 recursion levels';
my $res = ToIntegrationWorkflowCode($cmd, 'Raku::Ecosystem')
```

```raku
use MONKEY-SEE-NO-EVAL;
say $res.EVAL
```

```raku
my @commands =
        "compute the symbolic integral of Sin[x]*x^2 over 0 to 10",
        "integrate symbolically Sin[x]*x^2 over a b",
        "what is the integral of t^4+Log[t] for t in the interval 0 to a",
        "compute the numerical integral of Sin[x] for x over 0 to 10",
        "integrate numerically Sin[x^2]+5 in 0 to 10",
        "integrate numerically 1/x+Sqrt[x] in the interval 4 and 10",
        "integrate x^2+5",
        "integrate Exp[-x] over R+",
        "compute the integral of Log[x] of x from 0 to 10",
        "integrate 1/(x^2+y^2) over the domain [0.1, 10]x[0, 5]"
;

deduce-type(@commands)
```

```raku
my @res = @commands.map({
    use MONKEY-SEE-NO-EVAL;
    my $code = ToIntegrationWorkflowCode($_, 'Raku::Ecosystem');
    my $result = try $code.EVAL;
    if $! { $result = 'N/A' }
    %( command => $_, :$code, :$result )
});

deduce-type(@res):tally
```

```raku
#% html
my @field-names = <command code result>;
@res
==> to-html(:@field-names, align => 'left')
```
